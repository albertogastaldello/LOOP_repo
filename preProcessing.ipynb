```python
from utils import *
import pandas as pd
import os
import numpy as np
```

```python
# Load data 

file_path = os.path.join("tables", "exerciseTableForBN_python.parquet")
exerciseTable = pd.read_parquet(file_path)
```

```python
# Remove rows with missing values

exerciseTable.dropna(subset=["weight"], inplace=True)
exerciseTable.dropna(subset=["COB"], inplace=True)
exerciseTable.dropna(subset=["preExerciseRoc"], inplace=True)
exerciseTable.dropna(subset=["avgSessionsPerWeek"], inplace=True)
exerciseTable.dropna(subset=["diabetesDuration"], inplace=True)
```

```python
# map exercise modality
exerciseTable = map_exercise_modality(exerciseTable, activity_col_name="CleanActivityName")
```

```python
# Define columns to remove(variables never used as features)

colsToRemove = ["PtID", "DeviceDtTm", "UTCDtTm", "ExerciseName", "DistanceValue", "DistanceUnits", "DurationUnits", "EnergyValue",
                 "EnergyUnits", "TmZnOffset", "CleanActivityName", "TimeSinceLastBolus", "LastBolus", 
                 "TimeSinceLastBasal", "LastBasal"]

exerciseTable.drop(columns=colsToRemove, inplace=True)
```

```python
exerciseTable.to_csv(os.path.join("tables", "exerciseTableForBN_python_processed.csv"), index=False)
```
