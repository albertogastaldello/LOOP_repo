```python
import pandas as pd
import plotly.express as px
import textwrap

# 1. Define your exact mapping
modality_mapping = {
    'Walking': 'Aerobic', 'Cycling': 'Aerobic', 'Running': 'Aerobic',
    'Elliptical': 'Aerobic', 'Stairs': 'Aerobic', 'Swimming': 'Aerobic',
    'Hiking': 'Aerobic', 'StairClimbing': 'Aerobic', 'Rowing': 'Aerobic',
    'PaddleSports': 'Aerobic', 'Step Training': 'Aerobic',
    
    'Traditional Strength Training': 'Anaerobic', 'Functional Strength Training': 'Anaerobic',
    'Core Training': 'Anaerobic', 'Pilates': 'Anaerobic', 'Barre': 'Anaerobic',
    
    'High Intensity Interval Training': 'Mixed', 'Cross Training': 'Mixed',
    'Kick Boxing': 'Mixed', 'Mixed Cardio': 'Mixed', 'Mixed Metabolic Cardio Training': 'Mixed',
    'Dance': 'Mixed', 'Hockey': 'Mixed', 'Soccer': 'Mixed', 'Football': 'Mixed', 
    'Volleyball': 'Mixed', 'Racquetball': 'Mixed', 'Climbing': 'Mixed',
    'Downhill Skiing': 'Mixed', 'Snow Sports': 'Mixed', 'Snowboarding': 'Mixed',
    'Play': 'Mixed', 'Golf': 'Mixed',
    
    'Yoga': 'Other', 'Mind And Body': 'Other', 'Preparation And Recovery': 'Other', 
    'Other Activity': 'Other'
}

# 2. Convert to a Pandas DataFrame & Apply Text Wrapping
df = pd.DataFrame(list(modality_mapping.items()), columns=['Activity', 'Modality'])

# Use textwrap to break lines intelligently at ~12 characters, and join them with an HTML line break
df['Activity_Wrapped'] = df['Activity'].apply(lambda x: '<br>'.join(textwrap.wrap(x, width=12)))

# Add a weight column of 1 so every activity gets an equal-sized box
df['Weight'] = 1 

# 3. Build the Nested Treemap
# Notice we are now using the 'Activity_Wrapped' column for the path!
fig = px.treemap(
    df, 
    path=[px.Constant("Exercise Modalities"), 'Modality', 'Activity_Wrapped'], 
    values='Weight',
    color='Modality',
    color_discrete_map={
        'Aerobic': '#3498db',   
        'Anaerobic': '#e74c3c', 
        'Mixed': '#9b59b6',     
        'Other': '#2ecc71',
        'Exercise Modalities': '#ecf0f1' 
    }
)

# 4. Format for Presentation Legibility
fig.update_traces(
    textinfo="label",
    textfont=dict(size=16, family="Arial Black"), # Font size is now much more stable
    marker=dict(line=dict(color='white', width=2))
)

fig.update_layout(
    title=dict(
        text="Physiological Mapping of Exercise Modalities", 
        font=dict(size=24, family="Arial"),
        y=0.96, x=0.5, xanchor='center', yanchor='top'
    ),
    margin=dict(t=60, l=10, r=10, b=10) 
)

# 5. Display and Save
fig.show()
```

```python
import plotly.express as px
import os

file_path = os.path.join("tables", "exerciseTableForBN_python_processed.csv")
exerciseTable = pd.read_csv(file_path)

# 1. Define the Color Palette (Consistent with your Treemap)
color_palette = {
    'Aerobic': '#3498db',   
    'Anaerobic': '#e74c3c', 
    'Mixed': '#9b59b6',     
    'Other': '#2ecc71'
}

# 2. Create the Scatter Plot using your REAL dataframe
fig = px.scatter(
    exerciseTable,                  # <--- Your actual dataframe
    x="DurationValue",              # <--- Your actual duration column
    y="MET",                        # <--- Your actual MET column
    color="ExerciseModality",       # <--- Your new mapped modality column
    color_discrete_map=color_palette,
    opacity=0.5,                    # Keeps the 50% transparency for density
    marginal_x="histogram",         # Histograms on the top
    marginal_y="histogram",         # Histograms on the right
    hover_data=['ExerciseModality'] # Tooltip info
)

# 3. Styling the Chart for your Presentation
fig.update_traces(
    marker=dict(size=6, line=dict(width=0)), selector=dict(type='scatter') # Clean, borderless dots
)

fig.update_layout(
    title=dict(
        text="Metabolic Load vs. Duration across Real-World Exercise Sessions", 
        font=dict(size=22, family="Arial"),
        y=0.98, x=0.5, xanchor='center', yanchor='top'
    ),
    xaxis_title="Duration (Minutes)",
    yaxis_title="Intensity (METs)",
    plot_bgcolor='white',  
    legend=dict(
        title="Modality",
        orientation="h",   
        yanchor="bottom", y=-0.3, xanchor="center", x=0.5,
        font=dict(size=14)
    ),
    margin=dict(t=60, l=60, r=40, b=80)
)

# Add gridlines for readability
fig.update_xaxes(showgrid=True, gridwidth=1, gridcolor='#f0f0f0')
fig.update_yaxes(showgrid=True, gridwidth=1, gridcolor='#f0f0f0')

# 4. Display and Save
fig.show()
```
